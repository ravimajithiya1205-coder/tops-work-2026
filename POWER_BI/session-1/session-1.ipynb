<!-- The Power BI Desktop interface -->

1. Report View
--> Report View is used to create and design reports using charts, tables, cards, and other visualizations.

2. Data View
--> Data View displays the data imported into Power BI in a table format. It helps users inspect rows, columns, and values.

3. Model View
--> Model View shows the relationships between different tables. It helps users manage and understand the data model.

4. Fields Pane
--> The Fields Pane displays available tables and their columns. Users can select or drag fields into visuals.

5. Visualizations Pane
-->	The Visualizations Pane contains different types of charts, tables, and graphs used to represent data visually.

6. Filters Pane
--> The Filters Pane allows users to filter data based on selected conditions for a visual, page, or entire report.

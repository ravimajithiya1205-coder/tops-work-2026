S1. Python Data Types and Discount Calculation
The order total 499.50 has the float data type because it contains a decimal value. The delivery distance 7.3 also has the float data type. The payment method 'UPI' has the str (string) data type because it stores text, and the restaurant's order-accepting status True has the bool (Boolean) data type because it represents a true or false condition.

If the order total is incorrectly stored as an integer, the decimal part may be lost. For example, converting 499.50 to an integer results in 499. A 10% discount calculated on 499.50 is 49.95, but calculating it on 499 gives 49.9. This can produce an inaccurate result, so a float should be used when the order amount contains decimal values.

S2. List of Tuples vs Dictionary
A list of tuples stores each dish's name, price, and category together, such as ('Paneer Burger', 180, 'Snacks'). To find a dish's price, the program must search through the list and compare dish names until it finds a match. This takes linear time, or \(O(n)\), in the worst case.

A dictionary can store dish names as keys and their details or prices as values, such as {'Paneer Burger': 180, 'Masala Dosa': 90}. A dictionary usually provides faster average lookup, taking \(O(1)\) time, and its access is more readable because the dish name directly identifies the corresponding value. One limitation of a list of tuples is that the program must search manually for a dish, whereas a dictionary allows direct access by name.

S3. Named Function vs Lambda Function
A named function defined with def is more appropriate for the delivery fee calculator because it must handle multiple conditions. It needs to check whether the order value is at least Rs 500 and, if not, determine whether the delivery distance is 5 km or less or greater than 5 km. A named function makes these conditions easier to read, test, debug, and maintain.

A lambda function would be useful for a simple operation inside the same app, such as sorting orders by their total amount. For example, sorted(orders, key=lambda order: order['total']) sorts the orders by their total values without requiring a separate named function.
S4. Writing and Reading an Order Using JSON
To save an order, Python first imports the json module and creates an order dictionary containing the order details. It then opens a JSON file in write mode using open() and uses json.dump() to write the dictionary to the file. To reload the order, the program opens the file in read mode and uses json.load() to convert the stored JSON data back into a Python dictionary.

If the file does not exist when the program attempts to read it, Python raises a FileNotFoundError. The program should catch this exception and initialize the order history as an empty list or create an initial file so the app can start normally on its first run.

Example:

```python
import json

order = {
    "order_id": 101,
    "restaurant": "Domino's",
    "total": 499.50,
    "status": "Delivered"
}

# Write the order to a JSON file
with open("orders.json", "w") as file:
    json.dump(order, file, indent=4)

# Read the order from the JSON file
try:
    with open("orders.json", "r") as file:
        saved_order = json.load(file)
    print(saved_order)

except FileNotFoundError:
    saved_order = {}
    print("Order history not found. Starting with empty data.")
```

S5. Why a Class Is Better for Delivery Tracking
A class is a better design choice because it groups related delivery information and actions into one reusable structure. Each delivery object can store the rider's name, GPS location, assigned order ID, and current status. It can also provide methods to update the location and mark the delivery as complete, keeping the data and the operations that use it together.

This design applies encapsulation, which groups data and methods in one class and can control how the delivery information is updated. It also applies abstraction, which allows other parts of the app to use simple methods such as update_location() or complete_delivery() without needing to know all the internal implementation details. Additionally, inheritance could be used if the app later needs specialized delivery types that share common tracking features.


S6. Handling Invalid Input with try-except
I would use a try-except block to handle the ValueError raised when Python cannot convert the user's input into a number. I would place the input conversion inside the try block and catch ValueError in the except block. To let the user try again instead of terminating the program, I would place this logic inside a loop that continues until valid input is entered.

Example:

```python
while True:
    try:
        order_amount = float(input("Enter order amount: Rs "))
        if order_amount < 0:
            print("Order amount cannot be negative.")
            continue

        print("Order Amount:", order_amount)
        break

    except ValueError:
        print("Invalid input. Please enter a numeric amount.")
```

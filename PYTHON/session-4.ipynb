Given the string 'Apple iPhone 14 Pro Max', use string slicing to extract and print only the brand name and the model (i.e., 'Apple' and 'iPhone 14 Pro Max') separately.<br><br><em><strong>Hint:</strong> Use split() to help find the split point, then use slicing for the substrings.</em>

Suppose you have a list of messy product names: [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']. Write code to clean each name (remove spaces, replace hyphens with spaces, and make the brand title case) and print the cleaned list.<br><br><em><strong>Constraint:</strong> Use at least three string methods from this session.</em>

```python
# 1. Uppercase and lowercase

product_name = input("Enter product name: ")

print("Uppercase:", product_name.upper())
print("Lowercase:", product_name.lower())
```

```python
# Clean brand name 

def clean_brand_name(name):
    name = name.strip()
    name = name.replace("-", " ")
    return name


result = clean_brand_name(" oneplus-Nord ")

print(result)
```

```python
# 3. Brand and model using slicing

product = "Apple iPhone 14 Pro Max"

words = product.split()

brand = words[0]

model = product[len(brand) + 1:]

print("Brand:", brand)
print("Model:", model)
```

```python
# 4. Format product display

def format_product_display(name, price):
    return name + " - ₹" + str(price)


result = format_product_display("Boat Earbuds", 1299)

print(result)
```

```python
# 5. Clean messy product names

products = [
    " mi-Band 5 ",
    " SAMSUNG-Galaxy ",
    " realme-Book "
]

cleaned_products = []

for product in products:

    # Remove leading/trailing spaces
    product = product.strip()

    # Replace hyphen with space
    product = product.replace("-", " ")

    # Convert to title case
    product = product.title()

    cleaned_products.append(product)


print(cleaned_products)
```

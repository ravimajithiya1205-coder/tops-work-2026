```python
# 1. Extract Valid Indian Phone Numbers Using re.findall()

import re

text = """
OLX contact: 9876543210
Price: 15000
WhatsApp: 8765432109
Invalid number: 6123456789
Contact: 7654321098
Another number: 1234567890
"""

pattern = r'\b[789]\d{9}\b'

phone_numbers = re.findall(pattern, text)

print(phone_numbers)
```

```python
# 2. Check for a Date Using re.search()

import re

def check_date(text):
    pattern = r'\b\d{2}/\d{2}/\d{4}\b'

    if re.search(pattern, text):
        return True
    else:
        return False


print(check_date("Order delivered on 25/06/2024"))
print(check_date("Order delivered yesterday"))
```

```python
# 3. Extract Email Addresses Using re.findall()

import re

text = """
Zomato review:
Contact me at rahul123@gmail.com
Restaurant email: support@food.in
Invalid email: user@
Feedback: amit.shah@yahoo.com
"""

pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(pattern, text)

print(emails)
```

```python
# 4. Mask Phone Numbers Using Only re.sub()

import re

text = "Payment contact: 9876543210"

masked_text = re.sub(
    r'\b(\d{6})(\d{4})\b',
    r'******\2',
    text
)

print(masked_text)
```

```python
# 5. Match Flipkart-Style Order IDs Using re.search()

import re

order_strings = [
    "Your order ID is OD123456789012345000",
    "Order confirmed: OD987654321098765432",
    "Invalid order: OD12345",
    "Your package is on the way"
]

pattern = r'\bOD\d{18}\b'

for text in order_strings:
    match = re.search(pattern, text)

    if match:
        print("Order ID found:", match.group())
    else:
        print("No valid order ID found")
```

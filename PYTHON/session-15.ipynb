```python
# 1. Safe Division Using try-except

def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"


print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

```python
# 2. Zomato Rating System

try:
    reviews = int(input("Enter number of reviews: "))
    total_stars = int(input("Enter total stars: "))

    if reviews <= 0:
        print("Number of reviews must be greater than zero.")
    elif total_stars < 0 or total_stars > reviews * 5:
        print("Invalid total stars.")
    else:
        average_rating = total_stars / reviews
        print("Average rating:", average_rating)

except ValueError:
    print("Error: Please enter a valid number.")
```

```python
# 3. Calculate Playlist Duration Using a Custom Exception

class InvalidDurationError(Exception):
    pass


def get_playlist_duration(songs):
    total_seconds = 0

    for duration in songs:
        if duration < 0:
            raise InvalidDurationError(
                "Song duration cannot be negative."
            )

        total_seconds += duration

    return total_seconds / 60


durations = [180, 240, 300, 120]

try:
    total_minutes = get_playlist_duration(durations)
    print("Total playlist duration:", total_minutes, "minutes")

except InvalidDurationError as e:
    print("Error:", e)
```

```python
# 4. Flipkart Order Summary Using try-except-else-finally

try:
    price = float(input("Enter item price: Rs. "))
    quantity = int(input("Enter quantity: "))

    if price < 0 or quantity < 0:
        raise ValueError("Price and quantity cannot be negative.")

except ValueError as e:
    print("Invalid input:", e)

else:
    total_price = price * quantity
    print("Total price: Rs.", total_price)

finally:
    print("Thank you for shopping!")
```

```python
# 5. Divide Two Numbers and Handle Two Exceptions

try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")

except ValueError:
    print("Error: Please enter valid numeric values.")
```

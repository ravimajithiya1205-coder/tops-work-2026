```python
# 1. Recursive Function to Reverse a String

def reverse_string(s):
    if s == "":
        return s
    else:
        return reverse_string(s[1:]) + s[0]


print(reverse_string("hello"))
```

```python
# 2. Calculate Total Playlist Duration Using Recursion

def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    else:
        return durations[0] + sum_playlist_durations(durations[1:])


durations = [180, 240, 210, 300]

total = sum_playlist_durations(durations)

print("Total playlist duration:", total, "seconds")
```

```python
# 3. Local vs Global Variable

count = 10

def update_count():
    count = 5
    print("Inside:", count)


update_count()
print("Outside:", count)
```

```python
# 4. Recursive Function to Count Likes in Posts and Replies

def count_likes(posts):
    total = 0

    for post in posts:
        total += post["likes"]
        total += count_likes(post["replies"])

    return total


posts = [
    {
        "likes": 10,
        "replies": [
            {
                "likes": 5,
                "replies": [
                    {"likes": 2, "replies": []}
                ]
            },
            {
                "likes": 3,
                "replies": []
            }
        ]
    },
    {
        "likes": 20,
        "replies": [
            {"likes": 4, "replies": []}
        ]
    }
]

print("Total likes:", count_likes(posts))
```

```python
# 5. Lifetime of Local and Global Variables

app_status = "Offline"

print("Before function call:")
print("App status:", app_status)

def update_status():
    user_status = "Online"

    print("Inside function:")
    print("User status:", user_status)
    print("App status:", app_status)


update_status()

print("After function call:")
print("App status:", app_status)

# This would cause a NameError because user_status is local:
# print(user_status)
```

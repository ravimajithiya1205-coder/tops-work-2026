```python
# 1. Check IPL Ticket Eligibility

age = int(input("Enter your age: "))

if age >= 18:
    print("Eligible for IPL ticket booking")
else:
    print("Not eligible")
```

```python
# 2. Influencer Category

followers = int(input("Enter the number of followers: "))

if followers < 10000:
    print("Micro Influencer")
elif followers <= 100000:
    print("Rising Star")
else:
    print("Celebrity")
```

```python
# 3. Zomato Free Delivery

total = float(input("Enter your Zomato order total: "))

if total > 299:
    print("Apply Free Delivery")
elif total >= 200:
    print("Add more items for free delivery")
else:
    print("Delivery charges apply")
```

```python
# 4. Flipkart Cashback Using Nested If

cart_value = float(input("Enter your Flipkart cart value: "))
payment_method = input("Enter payment method (UPI, Card, Cash): ").upper()

if cart_value > 1000:
    if payment_method == "UPI":
        print("Eligible for 10% cashback")
    else:
        print("Eligible for 5% cashback")
else:
    print("No cashback")
```

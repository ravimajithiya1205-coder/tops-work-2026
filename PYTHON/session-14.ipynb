```python
# 1. Create my_playlist.txt and Write 5 Songs

with open("my_playlist.txt", "w") as file:
    file.write("Kesariya\n")
    file.write("Believer\n")
    file.write("Shape of You\n")
    file.write("Blinding Lights\n")
    file.write("Excuses\n")

print("Playlist file created successfully.")
```

```python
# 2. Read the Playlist and Print Songs in Uppercase

with open("my_playlist.txt", "r") as file:
    for song in file:
        print(song.strip().upper())
```

```python
# 3. Read IPL Match Scores Using CSV

import csv

with open("ipl_matches.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["Match"], "-", row["Winner"])
```

```python
# 4. Read Username and Followers from JSON

import json

with open("user_profile.json", "r") as file:
    data = json.load(file)

print("Username:", data["username"])
print("Followers:", data["followers"])
```

```python
# 1. Use iter() and next() to Print Movies

movies = [
    "Jawan",
    "Animal",
    "Dunki",
    "Pathaan",
    "Salaar"
]

movie_iterator = iter(movies)

print(next(movie_iterator))
print(next(movie_iterator))
print(next(movie_iterator))
print(next(movie_iterator))
print(next(movie_iterator))
```

```python
# 2. Print a Spotify Playlist Using enumerate()

songs = [
    "Kesariya",
    "Believer",
    "Shape of You",
    "Blinding Lights",
    "Excuses",
    "Perfect"
]

for position, song in enumerate(songs, start=1):
    print(f"{position}. {song}")
```

```python
# 3. Use zip() to Print a Zomato Menu

food_items = ["Pizza", "Burger", "Pasta", "Sandwich", "Dosa"]
prices = [250, 120, 180, 100, 80]

for food, price in zip(food_items, prices):
    print(f"{food} - ₹{price}")
```

```python
# 4. Create an Instagram Posts Generator

def insta_posts_generator(posts):
    for post in posts:
        yield post


posts = [
    "Enjoying my weekend!",
    "New memories ❤️",
    "Life is beautiful",
    "Keep smiling 😊"
]

post_generator = insta_posts_generator(posts)

try:
    while True:
        print(next(post_generator))

except StopIteration:
    print("All posts have been printed.")
```

```python
# 5. Create a Paytm Cashback Generator

def cashback_generator(transactions):
    for amount in transactions:
        cashback = amount * 0.05
        yield cashback


transactions = [100, 500, 1000, 250, 2000]

for cashback in cashback_generator(transactions):
    print(f"Cashback: ₹{cashback:.2f}")
```

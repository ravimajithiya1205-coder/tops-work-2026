```python
# 1. Create a Product class with a private attribute _price

class Product:
    def __init__(self, price):
        self.__price = price

    def display_price(self):
        print("Product Price:", self.__price)


p1 = Product(500)
p1.display_price()
```

```python
# 2. Add getter and setter methods with price validation

class Product:
    def __init__(self, price):
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative")
        self.__price = new_price


p1 = Product(500)

print("Current Price:", p1.get_price())

p1.set_price(800)
print("Updated Price:", p1.get_price())

# p1.set_price(-100)  # Raises ValueError
```

```python
# 3. Create a Playlist class using encapsulation

class Playlist:
    def __init__(self):
        self.__songs = []

    def add_song(self, song):
        self.__songs.append(song)
        print(song, "added to playlist")

    def remove_song(self, song):
        if song in self.__songs:
            self.__songs.remove(song)
            print(song, "removed from playlist")
        else:
            print(song, "not found in playlist")

    def get_songs(self):
        return self.__songs.copy()


my_playlist = Playlist()

my_playlist.add_song("Kesariya")
my_playlist.add_song("Apna Bana Le")
my_playlist.add_song("Heeriye")

print("Current Playlist:", my_playlist.get_songs())

my_playlist.remove_song("Apna Bana Le")

print("Updated Playlist:", my_playlist.get_songs())
```

```python
# 4. Create an abstract class PaymentMethod

from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class UPI(PaymentMethod):
    def pay(self, amount):
        print("Processing ₹", amount, "payment through UPI")


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print("Processing ₹", amount, "payment through Credit Card")


upi_payment = UPI()
upi_payment.pay(1500)

card_payment = CreditCard()
card_payment.pay(2500)
```
